```python
%matplotlib inline

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

diabetes = pd.read_csv('diabetes.csv')
diabetes.columns
```

```python
diabetes.head()
```

```python
print("Dimensi data diabetes: {}".format(diabetes.shape))
```

```python
# Mengidentifikasi jumlah pasien diabetes (1) dan tidak diabetes (0)
diabetes.groupby('Luaran').size()
```

```python

# Memisahkan data berdasarkan luaran
tidak_diabetes = diabetes[diabetes['Luaran'] == 0]
pasien_diabetes = diabetes[diabetes['Luaran'] == 1]

# Membuat histogram untuk pasien yang tidak diabetes
tidak_diabetes.hist(figsize=(12, 9), bins=10)
plt.suptitle('Distribusi Data Pasien Tidak Diabetes (Luaran = 0)',
             fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

# Membuat histogram untuk pasien yang diabetes
pasien_diabetes.hist(figsize=(12, 9), bins=10)
plt.suptitle('Distribusi Data Pasien Diabetes (Luaran = 1)',
             fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```
